```python
localization=True
sigmas = [False, True]
networks = [[32, 32], [128, 128], [512, 512], [1024, 1024], [16, 256], [16, 1024], [32, 1024], [8, 8, 8, 8], [16, 16, 16, 16], [8, 16, 32, 64], [64, 32, 16, 8], [128, 256, 512], [512, 256, 128], [1024, 1024, 1024, 1024]]
```

```python
def bootstrap():
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}; Seed {random_numbers[i]}.')
        he = HE(input_neurons=input_neurons, lr = lr, activation = activation, sigma = sigma, initial_sigma=initial_sigma, localization=localization)
        he.train(epochs)#, name=str(count)+ '-' + str(i+1) + '.png')s
```

```python
for input_neurons in networks:
    for sigma in sigmas:
        print(input_neurons, sigma)
        try:
            bootstrap()
        except Exception as e:
            print(e)
```
