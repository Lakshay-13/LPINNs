```python
n_runs = 5
random_numbers = np.random.randint(1, 100000000, size=n_runs)
```

```python
def bootstrap():
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}; Seed {random_numbers[i]}.')
        he = HE(input_neurons=input_neurons, lr = lr, activation = activation, mu = mu, sigma = sigma, initial_sigma=initial_sigma, localization=localization)
        he.train(epochs)#, name=str(count)+ '-' + str(i+1) + '.png')s
```

```python
input_neurons=[512, 512]
lr = 1e-2
activation = nn.Tanh
sigma = True
initial_sigma=0.1
localization=False
epochs = 3000
mu = True
sigma = True
```

```python
bootstrap()
```

```python
localization=True
networks = [[128, 128], [512, 512], [512, 64], [512, 16, 16], [256, 32, 16], [256, 32]]
```

```python
for input_neurons in networks:
        print(input_neurons, sigma)
        try:
            bootstrap()
        except Exception as e:
            print(e)
```
