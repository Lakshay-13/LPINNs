```python
from PIL import Image
import os

# Directory containing your PNG files
png_dir = 'temp'
# Output GIF file
output_gif = 'output.gif'

# Load all PNG files from the directory
images = []

# Collect file names in a list
file_names = [file_name for file_name in os.listdir(png_dir) if file_name.endswith('.png')]

# Sort file names based on the numeric value in the file name (assuming 'epoch-' prefix)
file_names.sort(key=lambda x: int(x.split('-')[1].split('.')[0]))

# Load images based on the sorted file names
for file_name in file_names:
    file_path = os.path.join(png_dir, file_name)
    images.append(Image.open(file_path))

# Save as GIF
if images:
    images[0].save(output_gif, save_all=True, append_images=images[1:], optimize=False, duration=300, loop=0)
```
