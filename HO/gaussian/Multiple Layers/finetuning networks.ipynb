```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
lr = 1e-2
activation = SineActivation
initial_sigma = 0.06
scaling = True
localization = True
sigma = True
epochs = 5000
```

```python
n_runs = 10
random_numbers = np.random.randint(1, 100000000, size=n_runs)
```

```python
def bootstrap():
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}:')
        oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime=u_0_prime, t_min=t_min, t_max=t_max, 
                        train_batch_size=train_batch_size, valid_batch_size=valid_batch_size, input_neurons=input_neurons,
                        lr=lr, activation=activation, localization=localization, scaling=scaling, 
                        sigma=sigma, initial_sigma=initial_sigma)
        #print(oscillator.solver.nets)
        oscillator.train(epochs)#, name=str(count)+ '-' + str(i+1) + '.png')s
```

```python
networks = [[8,8,8], [16, 16, 16], [32, 32, 32], [8, 16, 32], [16, 32, 64], [8, 16, 32, 64], [8, 16, 32, 64, 128], [8, 8, 8, 8, 8, 8], [8, 8, 8, 8, 8], [16, 16, 16, 16, 16], [32, 32, 32, 32]]
```

```python
for input_neurons in networks:
    print(input_neurons)
    try:
        bootstrap()
    except Exception as e:
        print(e)
```
