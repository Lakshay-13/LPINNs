```python
from harmonic_oscillator import *
```

```python
class SineActivation(nn.Module):
    def forward(self, input):
        return torch.sin(input)
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
lr = 1e-2
activation = SineActivation
input_neurons = [1024]
initial_sigma, sigma_min, sigma_max = [None, 1.5, 2.5]
scaling = True
```

```python
n_runs = 3
random_numbers = np.random.randint(1, 100000000, size=n_runs)
```

```python
def bootstrap(count):
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}:')
        oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                    train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                    sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation,initial_sigma=initial_sigma, sigma_min=sigma_min, 
                    sigma_max=sigma_max, localization=localization, scaling=True)
        oscillator.train(100, name=str(count)+ '-' + str(i+1) + '-1.png')
        oscillator.train(900, name=str(count)+ '-' + str(i+1) + '-2.png')
        oscillator.train(4000, name=str(count)+ '-' + str(i+1) + '-3.png')
        oscillator.train(5000, name=str(count)+ '-' + str(i+1) + '-4.png')
```

```python
def rendering_heatmaps(count):
    oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                    train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                    sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation,initial_sigma=initial_sigma, sigma_min=sigma_min, 
                    sigma_max=sigma_max, localization=localization, scaling=True)
    flag = 1
    for i in range(100)
        oscillator.train(100, name=str(count)+ '-' + str(i+1) + f'-{flag}.png')
        flag += 1
```

```python
localization = True
mu = False
sigma = False
rendering_heatmaps('mu_false_sigma_false')
```
