```python
from harmonic_oscillator import *
```

```python
class SineActivation(nn.Module):
    def forward(self, input):
        return torch.sin(input)
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
lr = 1e-2
activation = SineActivation
initial_sigma, sigma_min, sigma_max = [0.06, None, None]
scaling = True
localization = True
input_neurons = [16, 256]
mu = False
sigma = False
```

```python
oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation,initial_sigma=initial_sigma, sigma_min=sigma_min, 
                sigma_max=sigma_max, localization=localization, scaling=True)
```

```python
oscillator.train(500)
```

```python
solver = oscillator.solver
```

```python
t_vals = torch.linspace(solver.t_min, solver.t_max, 256, requires_grad = True)
residuals = solver.get_residuals(t_vals)
```

```python
derivatives = []
for res in residuals:
    derivatives.append(diff(res**2, solver.best_nets[0].layers[0].weight,shape_check=False))
```
