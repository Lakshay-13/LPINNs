```python
from harmonic_oscillator import *
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
mu=False
sigma=True
input_neurons=1024
lr = 1e-2
```

```python
n_runs = 5
```

```python
random_numbers = np.random.randint(0, 100, size=n_runs)
```

```python
for i in range(len(random_numbers)):
    torch.manual_seed(random_numbers[i])
    print(f'Run {i + 1}:')
    oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                sigma=sigma, input_neurons=input_neurons, lr = lr)
    oscillator.train(100)
    oscillator.train(900)
    oscillator.train(4000)
    oscillator.train(5000)
```
