```python
import time
start = time.time()
```

```python
from harmonic_oscillator import *
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=1024
valid_batch_size=64
input_neurons=[64, 64]
lr = 1e-2
activation = nn.SiLU
```

```python
n_runs = 5
```

```python
random_numbers = np.random.randint(0, 100, size=n_runs)
```

```python
def bootstrap():
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}:')
        oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                    train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                    sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation, initial_sigma=initial_sigma)
        oscillator.train(100)
        oscillator.train(900)
        oscillator.train(4000)
        oscillator.train(5000)
```

```python
sigmas = np.linspace(0,np.pi, 6)
```

```python
mu=False
sigma=True
for j in sigmas:
    print(f'Sigma: {j}')
    initial_sigma = j
    bootstrap()
```

```python
mu=True
sigma=True
for j in sigmas:
    print(f'Sigma: {j}')
    initial_sigma = j
    bootstrap()
```
