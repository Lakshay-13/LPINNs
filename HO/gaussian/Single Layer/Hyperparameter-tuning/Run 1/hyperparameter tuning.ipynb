```python
import time
start = time.time()
```

```python
from harmonic_oscillator import *
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
lr = 1e-2
activation = nn.SiLU
```

```python
n_runs = 5
```

```python
random_numbers = np.random.randint(1, 100000000, size=n_runs)
```

```python
def bootstrap(count):
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}:')
        oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                    train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                    sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation,initial_sigma=initial_sigma, sigma_min=sigma_min, sigma_max=sigma_max)
        oscillator.train(100, name=str(count)+ '-' + str(i+1) + '-1.png')
        oscillator.train(900, name=str(count)+ '-' + str(i+1) + '-2.png')
        oscillator.train(4000, name=str(count)+ '-' + str(i+1) + '-3.png')
        oscillator.train(5000, name=str(count)+ '-' + str(i+1) + '-4.png')
```

```python
params_networks = [[32, 32, 32], [32, 32], [1024, 32, 32], [1024, 32], [256, 64, 32], [256, 32, 32], [128, 128, 128], [128, 32, 32]]
params_locals = [[None, None, None], [0.3, None, None], [0.7, None, None], [1.0, None, None], [1.5, None, None], [2.0, None, None], [3.0, None, None],
                [None, 0.3, 0.7], [None, 1.0, 1.5], [None, 2.0, 3.0]]
```

```python
count = 1
for j in params_networks:
    input_neurons=j
    for i in params_locals:
        initial_sigma, sigma_min, sigma_max = i
        print(f'Sigmas trainable; Network: {j}; Initial Sigma: {i[0]}, Sigma Range: {i[1]} - {i[2]}.')
        mu=False
        sigma=True
        try:
            bootstrap(count)
            count=count+1
        except Exception as e:
            print(e)
        print(f'Mus and Sigmas trainable; Network: {j}; Initial Sigma: {i[0]}, Sigma Range: {i[1]} - {i[2]}.')
        mu=True
        sigma=True
        try:
            bootstrap(count)
            count=count+1
        except Exception as e:
            print(e)
```
