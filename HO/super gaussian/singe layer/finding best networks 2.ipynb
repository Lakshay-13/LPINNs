```python
class SineActivation(nn.Module):
    def forward(self, input):
        return torch.sin(input)
```

```python
t_0=0.0
u_0=0.0
u_0_prime = 1.0
t_min=0.0
t_max=4*np.pi
train_batch_size=256
valid_batch_size=64
lr = 1e-2
activation = SineActivation
initial_sigma, sigma_min, sigma_max = [0.065, None, None]
scaling = True
localization = True
mu = False
sigma = False
```

```python
n_runs = 10
random_numbers = np.random.randint(1, 100000000, size=n_runs)
```

```python
def bootstrap():
    for i in range(len(random_numbers)):
        torch.manual_seed(random_numbers[i])
        print(f'Run {i + 1}:')
        oscillator = HO(t_0=t_0, u_0=u_0, u_0_prime = u_0_prime, t_min=t_min, t_max=t_max, 
                    train_batch_size=train_batch_size, valid_batch_size=64, mu=mu, 
                    sigma=sigma, input_neurons=input_neurons, lr = lr, activation=activation,initial_sigma=initial_sigma, sigma_min=sigma_min, 
                    sigma_max=sigma_max, localization=localization, scaling=True)
        #print(oscillator.solver.nets)
        oscillator.train(5000)#, name=str(count)+ '-' + str(i+1) + '.png')s
```

```python
networks = [[2048], [128, 512], [1024, 1024]]
sigmas = np.append(np.linspace(0, 0.1, 11), np.linspace(2, 10, 9))
```

```python
for initial_sigma in sigmas:
    for input_neurons in networks:
        print(input_neurons, initial_sigma)
        try:
            bootstrap()
        except Exception as e:
            print(e)
```
